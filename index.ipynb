## Option B — run the *entire* series in one runtime

If you just want everything executed end-to-end in a single Colab session (≈5–8 min, CPU),
run the two cells below instead of opening chapters individually. This clones the repo, installs
dependencies + Tesseract, then executes all 15 notebooks in order and writes `VALIDATION_REPORT.md`.
The executed notebooks (with outputs) are saved under `notebooks/` in this runtime — open them
from the **Files** panel on the left.

```python
# 1. Clone + install (Colab). A PRIVATE repo is cloned using a GitHub token read from
#    Colab Secrets (key 'GITHUB_TOKEN') — add it via the key icon in the left sidebar
#    (scope: repo, enable Notebook access). The token is never printed.
import os, sys, subprocess
REPO_URL = "https://github.com/rsalehin/patent-rag-masterclass"
target = "/content/patent-rag-masterclass"
if not os.path.isdir(target):
    token = None
    try:
        from google.colab import userdata
        token = userdata.get("GITHUB_TOKEN")
    except Exception:
        token = os.environ.get("GITHUB_TOKEN")
    url = REPO_URL.replace("https://github.com/", f"https://{token}@github.com/") if token else REPO_URL
    r = subprocess.run(["git", "clone", "--depth", "1", url, target],
                       stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    if r.returncode != 0:
        raise RuntimeError("Clone failed. Add a Colab secret named GITHUB_TOKEN (repo scope, "
                           "Notebook access enabled), or make the repo public. See README.md.")
os.chdir(target)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
subprocess.run(["apt-get", "install", "-y", "-q", "tesseract-ocr"], check=False)
print("ready in", os.getcwd())
```

```python
# 2. Execute the whole series fresh, in order → VALIDATION_REPORT.md
import subprocess, sys
subprocess.run([sys.executable, "scripts/validate_all.py"], check=False)
print(open("VALIDATION_REPORT.md", encoding="utf-8").read())
```
